# Importar Datos de Ocupados

Importación de datos desde GitHub y almacenamiento en catálogo Unity Catalog como tabla Delta.

```python
import pandas as pd
import requests
from io import StringIO

# Leer datos desde GitHub
url = "https://raw.githubusercontent.com/niconomist98/DataAnalyticsUQ/202602/DATOS/OCUPADOS.CSV"
response = requests.get(url)
response.encoding = 'utf-8'

# Crear DataFrame de pandas
df = pd.read_csv(StringIO(response.text),sep=';')

# Mostrar información del DataFrame
print(f"Dimensiones: {df.shape}")
print(f"\nPrimeras filas:")
display(df.head())
# Convertir a Spark DataFrame
df_spark = spark.createDataFrame(df)

# Guardar como tabla Delta en el catálogo
# Ajusta el nombre de la tabla según tu catálogo y esquema
table_name = "default.electiva_ocupados"
df_spark.write.format("delta").mode("overwrite").saveAsTable(table_name)

print(f"\nTabla guardada exitosamente como: {table_name}")
```
